### The Problem with Notebook Conflicts

Jupyter Notebooks are JSON files. Without nbdime, conflicts look like this:

```json
<<<<<<< HEAD
{"cell_type": "code", "execution_count": 42, "source": [...]}
=======
{"cell_type": "code", "execution_count": 15, "source": [...]}
>>>>>>> origin/main
```

**Unreadable and dangerous!** One wrong edit breaks the entire notebook.

### The Solution: nbdime

**nbdime** provides a visual interface to resolve conflicts cell-by-cell, keeping your notebook structure intact!

## Step 7: Resolve the Conflict with nbdime!

Now comes the magic! We'll use nbdime's graphical merge tool to resolve the conflict.

**Run the cell below to launch nbdime's merge interface:**

````python
# Launch nbdime's graphical merge tool
!git -C {BASE_PATH}  mergetool --tool=nbdime
````

### What to Do in the Merge Interface

The nbdime web interface will show you three versions:

1. **Base** - The original version before either change
2. **Local** - YOUR version (what you just committed)
3. **Remote** - Your TEAMMATE's version (from the remote repository)

**For each conflicting cell:**
- Click on the version you want to keep, OR
- Manually edit to combine both changes
- Make sure the final version makes sense!

**After resolving in the interface:**
- Save the merged notebook
- Close the browser tab
- Continue to Step 8 below

## Step 8: Complete the Merge

After resolving the conflict in the nbdime interface, complete the merge:

````python
# Stage the resolved notebook and complete the merge
!git -C {BASE_PATH} add --all
!git -C {BASE_PATH} commit -m "Resolved merge conflict using nbdime"

print("\n Merge conflict resolved successfully!")
````
